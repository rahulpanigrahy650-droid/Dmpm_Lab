```python
# mean without using any function
x = list(map(int,(input("enter x :").split(","))))
sum = 0
for i in x:
  sum+=i
n = len(x)

mean = sum/n

print(f" mean = {mean}")
```

```python
# mean using function(library)
import numpy as np

x = np.array(list(map(int,input("enter elements : ").split(","))))

print(f"mean = {np.mean(x)}")
```

```python
# median without using function
x = list(map(int,input("enter elements :").split(",")))
x.sort()
n =  len(x)
if n%2 == 0:
  meadin =  (x[(n//2)-1] + x[n//2]) / 2
else:
  meadin = x[n//2]

print(f"median = {meadin}")
```

```python
# median using function
x = np.array(list(map(int,(input("enter x : ").split(",")))))
print(f"meadin = {np.median(x)}")
```

```python
# mode without using function
x = list(map(int,(input("enter x :").split(","))))
max_count = 0
for i in range(len(x)):
  count = 0
  for j in range(len(x)):
    if x[i] == x[j]:
      count+=1
    if count >max_count:
      max_count = count
      mode = x[i]
print(f"mode = {mode}")



```

```python
#  mode using library

import pandas as pd
df = pd.Series(input("enter x:").split(","))
print(f"mode = {df.mode()}")
```

```python
#variance using function

x = np.array(list(map(int,(input("enter x : ").split(",")))))
var = np.sum(np.square(x - np.mean(x))/len(x))
print(f"variance = {var}")
```

```python
# standard deviation using function

x = np.array(list(map(int,(input("enter x : ").split(",")))))
std = np.sqrt(np.sum((x - np.mean(x))**2/len(x)))
print(f"standard_devciation = {std}")
```

```python
# variance with out using function
x = list(map(int,(input("enter x :").split(","))))
sum = 0
for i in x:
  sum+=i
n = len(x)

mean = sum/n
sum = 0
for i in x :
  sum+=(i- mean)**2

variance = sum/len(x)

print(f"variance = {variance}")



```

```python
# Standard deviation without using sunction
x = list(map(float,(input("enter x :").split(","))))
sum = 0
for i in x:
  sum+=i
n = len(x)

mean = sum/n
sum = 0
for i in x :
  sum+=(i- mean)**2

standard_deviation = (sum/len(x))**0.5

print(f"standard_deviation = {standard_deviation}")
```
