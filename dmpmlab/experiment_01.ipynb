```python
# largest among 3 number
def largest(a,b,c):
    if a>b or a >c :
        print(f"largest = {a}")
    elif b>c:
        print(f"largest = {b}")
    else :
        print(f"largest = {c}")
a = int(input("enter a :"))
b = int(input("enter b :"))
c = int(input("enter c :"))
largest(a,b,c)
```

```python
# even or odd
def even_odd(n):
    if n%2 == 0:
        print(f"{n} is even number")
    else:
        print(f"{n} is not a even number")
n = int(input("enter n :"))
even_odd(n)
```

```python
# swap 2 number
def swap(a,b):
    print(f"before swap a= {a} , b = {b}")

    a = a+b
    b = a-b
    a = a-b

    print(f"after swap a = {a},b = {b}")

a = int(input("enter a = "))
b = int(input("enter b ="))
swap(a,b)
```

```python
#pallandromrme 

def pallandrome(n):
    left  = 0
    right = len(n) -1

    while left < right:
        if n[left] != n[right]:
            return False
        left +=1
        right -=1

    return True

n = input("enter a string :")
if pallandrome(n):
    print(f"{n} is pallandrome")
else:
    print("not pallandrome")
```

```python
# calculator
def calc(a, b, op):
    match op:
        case '+':
            return a + b
        case '-':
            return a - b
        case '*':
            return a * b
        case '/':
            return a / b
        case '//':
            return a//b
        case '%':
            return a%b
        case _:
            return "Invalid operator."


a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
op = input("Enter operator (+, -, *, /,//,%): ")

print(calc(a, b, op))
```

```python
# vowels in a string
def vowels(n):
    count =0
    for i in n:
        if i in "aeiouAEIOU":
            count+=1
    return count
n = input("enter a string :")
print(f"no of vowels in {n} = {vowels(n)}")
```

```python
# prime or not
def prime(n):
    count = 0
    for i in range(1,n+1):
        if n%i ==0 :
            count+=1


    if count == 2:
            print(f"{n} is prime")
    else:
            print(f" {n} is Not prime")

n = int(input('enter n :'))
prime(n)

```

```python
# add digit of a number
def add(n):
    sum = 0
    while (n>0):
        digit= n%10
        sum+=digit
        n//=10
    print(f"sum of digits  = {sum}")
n = int(input("enter n : "))
add(n)
```

```python
# reverse of num

def reverse(n):
    rev = 0
    while(n>0):
        digit  = n%10
        rev = rev*10+digit
        n//=10

    print(f"reverse  = {rev}")

n = int(input("enter n :"))
reverse(n)
```

```python
# factorial

def factorial(n):
    fact =1
    for i in range(1,n+1):
        fact*=i

    print(f"factorial = {fact}")
n = int(input("enter n :"))
factorial(n)
```
